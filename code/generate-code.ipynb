```python
import os
from pathlib import Path
import ollama

INPUT_DIR = "/kaggle/working/part_1/ml_filtered_repos"
OUTPUT_DIR = "output"
MODEL_NAME = "qwen2.5:14b"

def generate_python_code(data):
    response = ollama.generate(
        model=MODEL_NAME,
        prompt=f"""
You are an expert Python developer.

Convert the following text specification into complete Python code.

Requirements:
1. Generate clean, executable Python code.
2. Include necessary imports.
3. Maintain logical structure and readability.
4. Do not include explanations or markdown.
5. Output ONLY valid Python code.

Specification:
{data}
""",
        options={
            "num_ctx": 131072
        }
    )

    return response['response']


def process_txt_file(input_path, output_path):
    with open(input_path, "r", encoding="utf-8") as f:
        data = f.read()

    generated_code = generate_python_code(data)

    os.makedirs(os.path.dirname(output_path), exist_ok=True)

    with open(output_path, "w", encoding="utf-8") as f:
        f.write(generated_code)


def main():
    for root, dirs, files in os.walk(INPUT_DIR):
        for file in files:
            if file.endswith(".txt"):
                input_file_path = os.path.join(root, file)

                relative_path = os.path.relpath(input_file_path, INPUT_DIR)
                output_file_path = os.path.join(
                    OUTPUT_DIR,
                    os.path.splitext(relative_path)[0] + ".py"
                )

                print(f"Processing: {input_file_path}")

                process_txt_file(input_file_path, output_file_path)


if __name__ == "__main__":
    main()
```
